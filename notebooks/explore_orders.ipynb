```python
import sys; sys.path.insert(0, 'src')
import pandas as pd
from analysis import load, net_sales
df = load(); s = net_sales(df)
len(df), len(s)
```

## Which weekday sells most?

```python
wd = s.groupby(s['order_date'].dt.day_name())['revenue'].sum()
(wd / wd.sum() * 100).round(1).sort_values(ascending=False)
```

## Basket size by category

```python
s.groupby('category')['qty'].mean().round(2).sort_values(ascending=False)
```

## Price bands
Gift boxes dominate the top band.

```python
bands = pd.cut(s['unit_price'], [0, 200, 500, 1000, 5000], labels=['<200', '200-500', '500-1000', '1000+'])
pd.crosstab(bands, s['category'])
```
